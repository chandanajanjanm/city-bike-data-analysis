# NYC Citi Bike Data Science Project

## Project Objective

The goal of this project is to analyze NYC Citi Bike trip data
for January, February, and March 2025.

We will perform data collection, data cleaning,
feature engineering, exploratory data analysis,
and weather data integration.

## Data Sources

### Citi Bike Trip Data

- Source: NYC Citi Bike official trip data
- Period: January–March 2025
- Format: CSV files
- Data includes trip times, station information, and user type.

### Data Collection Process

1. Download monthly trip data.
2. Extract the downloaded files.
3. Load CSV files into Pandas.
4. Combine the monthly datasets.
5. Save the raw combined dataset.

## 1. Import Required Libraries

```python
import pandas as pd
import numpy as np
import requests
import os
import zipfile

print("All libraries imported successfully!")
```

## 2. Data Collection

We will collect NYC Citi Bike trip data
for January, February, and March 2025.

### 2.1 Download Citi Bike Data

```python
months = ["202501", "202502", "202503"]

for month in months:
    url = f"https://s3.amazonaws.com/tripdata/{month}-citibike-tripdata.zip"
    file_path = f"../data/raw/{month}-citibike-tripdata.zip"

    response = requests.get(url, timeout=30)
    response.raise_for_status()
    
    if response.status_code == 200:
        with open(file_path, "wb") as file:
            file.write(response.content)

        print(f"{month} downloaded successfully!")
    else:
        print(f"Failed to download {month}")
```

### 2.2 Extract Citi Bike Data

```python
months = ["202501", "202502", "202503"]

for month in months:
    zip_path = f"../data/raw/{month}-citibike-tripdata.zip"
    extract_folder = f"../data/raw/{month}"

    os.makedirs(extract_folder, exist_ok=True)

    with zipfile.ZipFile(zip_path, "r") as zip_ref:
        zip_ref.extractall(extract_folder)

    print(f"{month} extracted successfully!")
```

## 3. Data Loading

### 3.1 Load Citi Bike Data

```python
import glob

all_files = []

for month in ["202501", "202502", "202503"]:
    files = glob.glob(f"../data/raw/{month}/*.csv")
    all_files.extend(files)

print(f"Total CSV files found: {len(all_files)}")
```

### 3.2 Combine All CSV Files

```python
df_list = []

for file in all_files:
    df = pd.read_csv(file, low_memory=False)
    df_list.append(df)

combined_df = pd.concat(df_list, ignore_index=True)

print("Data combined successfully!")
print("Shape:", combined_df.shape)
```

### 3.3 Save Raw Combined Data

```python
import os

os.makedirs("../data/processed", exist_ok=True)

combined_df.to_csv(
    "../data/processed/citibike_raw_combined.csv",
    index=False
)

print("Raw combined dataset saved successfully!")
print("Shape:", combined_df.shape)
```
